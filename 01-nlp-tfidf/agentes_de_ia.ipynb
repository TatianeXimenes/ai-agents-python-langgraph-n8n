```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer # Transforma textos em dados numericos
```

```python
# Lista de frases
frases = [
    "Fernanda gosta de ir a praia. Creuza também gosta de praia.",
    "Fernanda também gosta de assistir futebol.",
    "José foi para a praia com Creuza e Fernanda."
]
```

```python
# Cria uma instância do CountVectorizer. A configuração específica lowercase=False impede que o algoritmo converta automaticamente todas as letras do seu texto para minúsculas antes de contar.
Bow = CountVectorizer(lowercase=False)
```

```python
# Ajustando e transformando as frases em uma matriz de contagem.
# fit: lê todas as frases, aprende o vocabulário e calcula o IDF de cada termo.
# transform: converte cada frase em um vetor de pesos TF-IDF.
# fit_transform faz as duas coisas de uma vez e devolve uma matriz esparsa (só guarda os valores diferentes de zero).
Bow_X = Bow.fit_transform(frases).toarray()

print(f"Vetor de representação BOW: \n{Bow_X}")
```

```python
# Criação do dataFrame com nomes das colunas
# cada linha é uma frase (na mesma ordem de frases);
# cada coluna é uma palavra do vocabulário, em ordem alfabética (get_feature_names_out() devolve esses nomes);
# cada célula é o peso TF-IDF daquela palavra naquela frase.
Bow_X_df = pd.DataFrame(data=Bow_X, columns=Bow.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import CountVectorizer # Também tokeniza e monta o vocabulário, mas devolve apenas a contagem bruta de cada palavra por documento (o “bag of words”), sem o peso IDF.
import pandas as pd
from nltk.downloader import download # O NLTK (Natural Language Toolkit) não vem com seus corpora e listas de palavras embutidos.
download('stopwords')
from nltk.corpus import stopwords
```

```python
# Definindo as stopwords
stop_words_portuguese = list(stopwords.words("portuguese"))
```

```python
list(stopwords.words("portuguese"))[:10]
```

```python
# Instanciando o CountVectorizer com n-grams de 1 a N
Bow = CountVectorizer(ngram_range=(1,1), # pode fazer combinações de palavras alterando o segundo valor
                        stop_words=stop_words_portuguese, 
                        strip_accents = 'unicode')
# stop_words=stop_words_portuguese: remove as palavras da lista do NLTK que você criou antes.
# strip_accents='unicode': remove acentos.
```

```python
# Ajustando e transformando as frases em uma matriz de contagem
Bow_X = Bow.fit_transform(frases).toarray()
```

```python
# Criando um DataFrame com os dados do Bow, uma tabela com uma linha por frase e uma coluna por termo do vocabulário.
Bow_X_df = pd.DataFrame(data=Bow_X, columns=Bow.get_feature_names_out())
```
