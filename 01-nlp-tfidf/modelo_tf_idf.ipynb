Rascunhos de aula sobre TF-IDF (Term Frequency Inverse Document Frequency), do curso 'Agentes de IA com python'

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

Esse import traz para o seu código a classe TfidfVectorizer do scikit-learn, que transforma uma coleção de textos em uma matriz numérica usando TF-IDF (Term Frequency – Inverse Document Frequency).

Cada palavra recebe um peso que combina duas coisas:
1. TF (frequência do termo): quantas vezes a palavra aparece naquele documento. Quanto mais aparece, maior o peso.
2. IDF (frequência inversa do documento): quão rara a palavra é no conjunto inteiro de documentos. Palavras que aparecem em quase todos os textos (como “de”, “que”, “o”) recebem peso baixo; palavras mais raras e distintivas recebem peso alto.

O resultado é que termos realmente característicos de um documento ganham destaque, enquanto palavras comuns perdem importância.

```python
# Exemplo clássico
docs = [
    "dados de saúde pública",
    "qualidade de dados é essencial",
    "saúde pública no Brasil",
]

vec = TfidfVectorizer()
X = vec.fit_transform(docs)

print(vec.get_feature_names_out())  # vocabulário aprendido
print(X.shape)                      # (3 documentos, N termos)
print(X.toarray())                  # pesos TF-IDF
```

Voltando ao contexto do curso

```python
# Lista de frases
frases = [
    "Fernanda gosta de ir a praia. Creuza também gosta de praia.",
    "Fernanda também gosta de assistir futebol.",
    "José foi para a praia com Creuza e Fernanda."
]
```

```python
# Criação de vetor IT-IDF com desativação da normalização. Por padrão, o scikit-learn normaliza cada linha (documento) para ter norma L2 igual a 1, o que deixa os valores sempre entre 0 e 1. Com norm=None, você vê os pesos TF-IDF brutos: TF (contagem) multiplicado pelo IDF, sem escalonamento. Isso é bom para estudar, porque dá para conferir a conta na mão.
Tfidf = TfidfVectorizer(norm=None)
```

```python
# Transformação das frases em matriz TF-IDF
# fit: lê todas as frases, aprende o vocabulário e calcula o IDF de cada termo.
# transform: converte cada frase em um vetor de pesos TF-IDF.
# fit_transform faz as duas coisas de uma vez e devolve uma matriz esparsa (só guarda os valores diferentes de zero).
Tfidf_x = Tfidf.fit_transform(frases).toarray()
print(f"Vetor de representação TF-IDT: \n{Tfidf_x}")
```

```python
# Criação do dataFrame com nomes das colunas
# cada linha é uma frase (na mesma ordem de frases);
# cada coluna é uma palavra do vocabulário, em ordem alfabética (get_feature_names_out() devolve esses nomes);
# cada célula é o peso TF-IDF daquela palavra naquela frase.
Tfidf_x = pd.DataFrame(data=Tfidf_x, columns=Tfidf.get_feature_names_out())
```

```python
from sklearn.feature_extraction.text import CountVectorizer # Também tokeniza e monta o vocabulário, mas devolve apenas a contagem bruta de cada palavra por documento (o “bag of words”), sem o peso IDF.
import pandas as pd
from nltk.downloader import download # O NLTK (Natural Language Toolkit) não vem com seus corpora e listas de palavras embutidos.
download('stopwords')
from nltk.corpus import stopwords
```

```python
# Definindo as stopwords brasileiras
stop_words_portuguese = list(stopwords.words("portuguese"))
```

```python
# Instanciando o TfidfVectorizer com n-grams de 1 a N
Tfidf = TfidfVectorizer(ngram_range=(1,3), stop_words=stop_words_portuguese,
                            strip_accents = 'unicode', max_features=10, norm='l1') 
# soma do vetor de cada frase vai ser 1 [melhor para semelhança de textos]; l2 é o 'padrão' para treinar ML
# ngram_range=(1,3): Pensando no vocabulário de saúde, ele inclui unigramas (“dados”), bigramas (“dados saúde”) e trigramas (“dados saúde pública”). Isso captura um pouco de contexto, mas o vocabulário cresce muito. Os n-gramas são formados depois da remoção das stop words, então “dados de saúde” vira o bigrama “dados saúde”.

# stop_words=stop_words_portuguese: remove as palavras da lista do NLTK que você criou antes.
# strip_accents='unicode': remove acentos.
# max_features=10: mantém só os 10 termos mais frequentes no corpus inteiro. 
# norm='l1': normaliza cada linha para que a soma dos valores absolutos seja 1.
```

```python
# Ajustando e transformando as frases em uma matriz de contagem
Tfidf_X = Tfidf.fit_transform(frases).toarray()
```

```python
# Criando um DataFrame com os dados do Tfidf, uma tabela com uma linha por frase e uma coluna por termo do vocabulário.
Tfidf_X_df = pd.DataFrame(data=Tfidf_X, columns=Tfidf.get_feature_names_out())
```
